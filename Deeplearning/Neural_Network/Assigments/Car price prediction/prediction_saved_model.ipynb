```python
import pandas as pd
import torch
from torch import nn
import matplotlib.pyplot as plt
import os
```

```python
## Load the model and other details
X_mean  = torch.load("./model/X_mean.pt")
X_std  = torch.load("./model/X_std.pt")
y_mean  = torch.load("./model/y_mean.pt")
y_std  = torch.load("./model/y_std.pt")
model = nn.Linear(2,1)
model.load_state_dict( torch.load("./model/model.pt", weights_only=True))
```

```python
X_pred = torch.tensor([[2020,10000],[2021,10000],[2022,10000]],dtype = torch.float32)
pred = model((X_pred-X_mean)/X_std)
print(pred, pred*y_std+y_mean)
```
